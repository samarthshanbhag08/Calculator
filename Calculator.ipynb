```python
def calc():
    import math
    n1 = int(input('Please input a number1: '))
    n2 = int(input('Please input a number2: '))
    op = 'something'
    while op not in ['+','-','/','*','log','sin','cos','tan','^']:
        op = input('Please choose your operator: ')
        if op == '+':
            print(n1+n2)
        elif op == '-':
            print(n1-n2)
        elif op == '/':
            print(n1/n2)
        elif op == '*':
           print(n1*n2)
        elif op == 'log':
            rs ='a'
            while rs not in ['10','e','N']:
                rs = input('Would you like to put the base as 10 or e(10,e,N): ')
                
            
        

            
            
            if rs == 'N':
                print(math.log(n1,n2))
            else:
                rs2 ='as'
                while rs2 not in [n1,n2]:
                    rs2 = int(input('For which number?: '))
                if rs == '10':   
                    print(math.log(rs2,10))
                elif rs == 'e':
                    print(math.log(rs2, math.e))
            
        elif op == 'sin':
            rs2 ='a'
            while rs2 not in [n1,n2]:
                    rs2 = int(input('For which number?: '))
            print(math.sin(rs2))
        elif op == 'cos':
            rs2 ='a'
            while rs2 not in [n1,n2]:
                    rs2 = int(input('For which number?: '))
            print(math.cos(rs2))
        elif op == 'tan':
            rs2 ='a'
            while rs2 not in [n1,n2]:
                    rs2 = int(input('For which number?: '))
            print(math.tan(rs2))
        elif op == '^':
            print(n1**n2)
   
    
```
